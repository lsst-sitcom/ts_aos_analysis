This notebook replicates what MTAOS does to derive the corrections. It can only run from the summit, as the quickLook collections are not available at usdf It's important to pass it te right configuration paths.

You will need ts_ofc (>4.2.0) and ts_aos_analysis (develop), lsst_distrib (w2025_38) set up in your environment.

```python
import logging
import numpy as np

import matplotlib.pyplot as plt
from lsst.daf.butler import Butler
from lsst.ts.aos.analysis import stateUtils, ofcUtils
from lsst.summit.utils.efdUtils import makeEfdClient

%matplotlib inline
%reload_ext autoreload
%autoreload 2
```

```python
butler = Butler(
    "LSSTCam",
    collections="LSSTCam/runs/quickLook",
    instrument="LSSTCam"
)
efd_client = makeEfdClient()
config_dir = '/home/gmegias/notebooks/lsst-ts/ts_config_mttcs/MTAOS/v8/ofc/' # need to reset this to your own directory
aos_fetcher = ofcUtils.AOSFetcher(butler, efd_client, config_dir=config_dir)
log = logging.getLogger("CamHexLut")

log.setLevel(logging.INFO) # comment this line out if you don't want to see all what MTAOS is doing
```

```python
# select the visit id you want to compute corrections for 
visitid = 2025091900249
dof_result = await aos_fetcher.get_mtaos_correction(visitid,  "0-9,10-16,30-34", 12, log=log)
```

```python
for i, s, in enumerate(dof_result):
    print(f" {i:2} {s:12.3f}")
```

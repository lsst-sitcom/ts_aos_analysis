This notebook plots Zernikes and AOS image quality as a function of sequence number.
It grabs data from the Butler, EFD, ConsDB, Gemini RINGSS, and Rubin DIMM.

```python
from lsst.ts.aos.analysis import NightlyAnalyzer
```

You set the `day_obs` when creating the analyzer.
You can also set `seq_min` and `seq_max` on creation if you do not want to wait for the analyzer to gather data for every exposure.

```python
analyzer = NightlyAnalyzer(20250422)
```

Once you have created the analyzer, use this line to update the database -- i.e., grab any new exposures since the last update

```python
analyzer.update()
```

Now we can create some plots to analyze the night.

First of all, the bottom of every plot contains a color bar that tells you which filter was used for each exposure.
The analyzer can print the color code for you:

```python
analyzer.plot_band_color_code()
```

Now let's plot the evolution of the image quality over the night

```python
fig, ax = analyzer.plot_image_quality()
```

Now let's plot the evolution of the AOS residual, including scatter-across-detectors, as well as the scatter-across-detectors for each Zernike in the final exposure.

However if we plot everything the plot will be way too crowded!
Let's just plot the last 20 exposures.

```python
fig, axes = analyzer.plot_history_resid(lookback=20)
```

Now we can plot the evolution of each Zernike

```python
fig, axes = analyzer.plot_history_zk(lookback=20)
```
